Bivariate and multivariate analyses investigating financial relationships between income, expenses, FOIR, scenarios, and loan eligibility.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../emi_prediction_dataset.csv', low_memory=False)

# Clean salary for plotting
df['monthly_salary'] = pd.to_numeric(df['monthly_salary'].astype(str).str.replace(r'(\..)+$', '', regex=True), errors='coerce')

plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='emi_eligibility', y='monthly_salary', palette='Set2')
plt.title('Monthly Salary Distribution across EMI Eligibility Classes')
plt.show()
```

```python
# Scenario vs Eligibility Crosstab
ct = pd.crosstab(df['emi_scenario'], df['emi_eligibility'], normalize='index') * 100
print('Eligibility % by Loan Scenario:')
print(ct.round(2))
```
