Formal assessment across Completeness, Uniqueness, Validity, Consistency, and Range Bound checks.


```python
import pandas as pd
import numpy as np

df = pd.read_csv('../emi_prediction_dataset.csv', low_memory=False)

# Completeness & Null Analysis
nulls = df.isnull().sum()
print('Columns with Missing Values:')
print(nulls[nulls > 0])
```

```python
# Categorical Consistency (e.g. gender labels)
print('Unique Gender Values:', df['gender'].value_counts())

# Outlier & Credit Score Domain Checks
print('Credit Score Outliers (< 300 or > 900):')
print('Under 300:', (df['credit_score'] < 300).sum())
print('Over 900:', (df['credit_score'] > 900).sum())
```
